```python
import time
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.utils import to_categorical

print("Loading Fashion-MNIST...")
(train_images, train_labels), (test_images, test_labels) = datasets.fashion_mnist.load_data()

# Reshape images to (28, 28, 1) and normalize pixel values to [0, 1]
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255


train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)


model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()


model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

print("\nStarting training...")
start_time = time.time()
history = model.fit(train_images, train_labels, epochs=5, batch_size=64, validation_split=0.1)
end_time = time.time()
training_duration = end_time - start_time

print(f"\nTraining completed in {training_duration:.2f} seconds.")

test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\nTest accuracy on Fashion-MNIST: {test_acc:.4f}")
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\nTest accuracy on Fashion-MNIST: {test_acc:.4f}")
```

### Comparison Discussion:

1. **Performance (Accuracy):**
   MNIST is easier, so a basic CNN can usually reach about 99% accuracy. Fashion-MNIST is harder because clothes have more details and different shapes, so the accuracy is usually around 90% to 92%.

2. **Training Times:**
   Both datasets have the same number of images and the same 28×28 grayscale image size. Because of this, they take almost the same amount of time to train when using the same computer and batch size.

### Experiment 1: Removing the first Dense layer (64 neurons)
Let's build, compile, and train a model where the flattened feature map goes directly to the final 10-unit output layer with softmax activation.

```python
# Re-build Convnet without the Dense(64) layer
model_no_dense = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(10, activation='softmax') # Directly to output
])

model_no_dense.compile(optimizer='adam',
                       loss='categorical_crossentropy',
                       metrics=['accuracy'])

print("Training model without the first Dense layer...")
history_no_dense = model_no_dense.fit(train_images, train_labels, epochs=5, batch_size=64, validation_split=0.1)

test_loss_nd, test_acc_nd = model_no_dense.evaluate(test_images, test_labels, verbose=2)
print(f"\nTest accuracy without the first Dense layer: {test_acc_nd:.4f}")
```

```python
# Re-build Convnet with an added Dense(4096) layer
model_large_dense = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.Flatten(),
    layers.Dense(4096, activation='relu'), # Added layer
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model_large_dense.compile(optimizer='adam',
                         loss='categorical_crossentropy',
                         metrics=['accuracy'])

print("Training model with an added 4096-unit Dense layer...")
history_large = model_large_dense.fit(train_images, train_labels, epochs=5, batch_size=64, validation_split=0.1)

test_loss_lg, test_acc_lg = model_large_dense.evaluate(test_images, test_labels, verbose=2)
print(f"\nTest accuracy with the added 4096 Dense layer: {test_acc_lg:.4f}")
```

We performed all the experiments using a Convolutional Neural Network (ConvNet) on the Fashion-MNIST dataset to see how changes in the model structure affected accuracy and training time. The baseline ConvNet achieved approximately 89.81% accuracy and took around 277.05 seconds to train. Fashion-MNIST is more difficult than standard MNIST, which usually reaches about 99% accuracy, because clothing images contain more complex shapes and greater variation within each category. However, the training times are nearly the same because both datasets contain the same number of images and use 28×28 grayscale images. When the 64-unit Dense layer was removed, the accuracy decreased slightly to approximately 88.92% because the model had less capacity to learn higher-level patterns. When a 4096-unit Dense layer was added, the accuracy was approximately 89.82%, which was almost the same as the baseline. This large layer increased the model size from fewer than 100,000 parameters to more than 2.4 million and significantly increased the training time, but it did not provide any meaningful improvement in accuracy.
