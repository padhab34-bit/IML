```python
import pandas as pd

url = 'https://gist.githubusercontent.com/sarchak/b87ad2be315ce05f7a047550646f3c41/raw/0c7d83608111f74f212398fc0d2a704e4f8dc499/diamonds.csv'
df = pd.read_csv(url, index_col=0)
```

```python
display(df.head(7))
```

```python
columns_to_describe = ['carat', 'depth', 'table', 'price', 'x', 'y', 'z']
df_reset = df.reset_index()
display(df_reset[columns_to_describe].describe())
```

```python
categorical_columns = ['cut', 'color', 'clarity']
display(df[categorical_columns].describe())
```

```python
print('Unique values for all columns in the DataFrame:')
for col in df.columns:
    print(f"  '{col}': {df[col].unique()}")
print(f"  'carat' (index): {df.index.unique().to_numpy()}")
```

```python
import matplotlib.pyplot as plt

%matplotlib inline

df_reset.hist(figsize=(15, 10))
plt.tight_layout()
plt.show()
```
